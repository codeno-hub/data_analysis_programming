```python
def process_marks():
    source = open("marks.txt", "r")
    destination = open("processed_results.txt", "w")

    for line in source:
        name, score = line.strip().split(",")
        score = float(score)
        new_value = score + 5
        destination.write(f"{name},{new_value:.1f}\n")

    source.close()
    destination.close()
    print("Processing complete. Results saved to processed_results.txt.")


if __name__ == "__main__":
    process_marks()
```
